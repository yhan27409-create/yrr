## 3. 훈련 통계로 표준화하고 원-핫으로 바꾸기

이때 평균과 표준편차는 **훈련 세트에서만** 구한다. 전체 표로 구하면 테스트 데이터의 정보가 훈련에 새어 들어가, 성능이 실제보다 좋게 나온다.

### 3.2 원-핫 인코딩

범주형 열 `origin`은 값 하나를 **열 세 개**로 바꾼다. 미국이면 `origin_1`만 1, 나머지는 0이다.

````python
sample = train_df[['origin']].head(4).copy()
sample['origin'] = pd.Categorical(sample['origin'], categories=[1, 2, 3])
print(pd.get_dummies(sample, columns=['origin']).astype(int))
````

`pd.Categorical`로 범주를 `[1, 2, 3]`으로 **고정**한다. 어느 세트에 우연히 유럽 차가 없어도 `origin_2` 열이 만들어져 세 세트의 열 개수가 같아진다.

### 3.3 함수 하나로 세 세트에 똑같이

표준화와 원-핫을 함수 하나에 넣고 세 세트에 **같은 함수**를 적용한다. 함수 안의 `mu`, `sd`는 훈련 세트의 값이다.

````python
def prep(df):
    out = df.copy()
    out[num_cols] = (out[num_cols] - mu) / sd
    out['origin'] = pd.Categorical(out['origin'], categories=[1, 2, 3])
    return pd.get_dummies(out, columns=['origin'])

train_p, val_p, test_p = prep(train_df), prep(val_df), prep(test_df)
print(train_p.shape, val_p.shape, test_p.shape)
print(train_p[num_cols].mean().round(4))
````

세 세트의 열이 **10개**로 같다. 훈련 세트의 숫자형 열 평균은 정확히 **0**이다. 검증과 테스트 세트는 훈련의 평균을 뺐으므로 0이 아니어도 된다.

## 5. 학습 함수와 평가 함수

### 5.1 한 에폭의 학습

실습 3의 다섯 줄을 **배치마다** 실행하고, 배치 손실의 평균을 돌려주는 함수를 만든다. `model.train()`은 모델을 학습 모드로 둔다.

````python
def train_one_epoch(model, loader, criterion, optimizer):
    model.train()
    total = 0.0
    for xb, yb in loader:
        optimizer.zero_grad()
        pred = model(xb)
        loss = criterion(pred, yb)
        loss.backward()
        optimizer.step()
        total += loss.item() * len(yb)
    return total / len(loader.dataset)
````

`loss.item() * len(yb)`는 배치 손실을 배치 크기만큼 되돌린 합이다. 마지막 배치가 작아도 데이터 한 건이 같은 무게를 갖도록 전체 건수로 나눈다.

### 5.2 평가는 기울기 없이

검증과 테스트에서는 가중치를 바꾸지 않는다. `model.eval()`로 평가 모드로 두고, `torch.no_grad()` 안에서 손실만 계산한다.

````python
def evaluate(model, loader, criterion):
    model.eval()
    total = 0.0
    with torch.no_grad():
        for xb, yb in loader:
            total += criterion(model(xb), yb).item() * len(yb)
    return total / len(loader.dataset)
````

`train()`과 `eval()`은 드롭아웃처럼 학습과 평가에서 다르게 동작하는 층을 전환한다. 지금 모델에는 그런 층이 없지만 항상 붙이는 습관을 들인다.

### 5.3 모델과 도구 준비

입력 9개를 받아 숫자 하나를 내는 모델이다. 은닉층 두 개를 둔다. 손실은 실습 3과 같은 `MSELoss`이고, 갱신 도구는 학습률을 스스로 조절하는 `Adam`을 쓴다.

````python
import torch.nn as nn

torch.manual_seed(42)
model = nn.Sequential(
    nn.Linear(9, 32),
    nn.ReLU(),
    nn.Linear(32, 16),
    nn.ReLU(),
    nn.Linear(16, 1),
)
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

print(evaluate(model, val_loader, criterion))
````

학습 전 검증 손실은 약 **613**이다. 연비의 제곱 단위이므로 수백이 나온다.

## 6. 학습하며 러닝커브 기록하기

### 6.1 에폭마다 훈련 손실과 검증 손실

에폭마다 훈련 함수와 평가 함수를 한 번씩 부르고 두 손실을 기록한다.

````python
history = {'train': [], 'val': []}

for epoch in range(200):
    tr = train_one_epoch(model, train_loader, criterion, optimizer)
    va = evaluate(model, val_loader, criterion)
    history['train'].append(tr)
    history['val'].append(va)

print("첫 에폭  훈련:", round(history['train'][0], 2), " 검증:", round(history['val'][0], 2))
print("마지막   훈련:", round(history['train'][-1], 2), " 검증:", round(history['val'][-1], 2))
print("최소 검증 손실:", round(min(history['val']), 2), " 에폭:", history['val'].index(min(history['val'])))
````

훈련 손실은 약 **600에서 4**로, 검증 손실은 약 **523에서 6**으로 줄었다. 그런데 마지막 에폭의 검증 손실이 최소 검증 손실(약 **4.8**)보다 **크다**. 어느 에폭 이후로는 더 돌려도 새 데이터에 대한 성능이 좋아지지 않았다.

### 6.2 러닝커브 그리기 — 제공 코드

````python
import matplotlib.pyplot as plt

best_epoch = history['val'].index(min(history['val']))

plt.plot(history['train'], label='train')
plt.plot(history['val'], label='validation')
plt.axvline(best_epoch, color='green', linestyle='--', label=f'best epoch {best_epoch}')
plt.yscale('log')
plt.xlabel('Epoch')
plt.ylabel('MSE')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
````

### 6.3 러닝커브 읽기

| 훈련 손실 | 검증 손실 | 진단 | 처방 |
|---|---|---|---|
| 높다 | 높다 | 과소적합 | 모델을 키우거나 더 학습한다 |
| 낮다 | 내려가다 올라간다 | 과적합 | 최적 지점에서 멈춘다 |
| 낮다 | 낮다 | 적절 | |

훈련 손실은 계속 내려가지만 검증 손실은 어느 지점부터 다시 오른다. 훈련 손실만 그리면 이 판단을 할 수 없다.

## 7. 최적 지점 저장과 조기 종료

### 7.1 가장 좋았던 가중치를 복사해 두기

학습을 다시 하되, 이번에는 검증 손실이 가장 낮았던 순간의 가중치를 복사해 둔다. `model.state_dict()`는 모델의 모든 가중치와 편향을 담은 사전이다. `copy.deepcopy`로 그 순간의 값을 따로 보관한다.

````python
import copy

torch.manual_seed(42)
model = nn.Sequential(
    nn.Linear(9, 32), nn.ReLU(),
    nn.Linear(32, 16), nn.ReLU(),
    nn.Linear(16, 1),
)
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

best_val, best_state, best_epoch = float('inf'), None, 0
for epoch in range(200):
    train_one_epoch(model, train_loader, criterion, optimizer)
    va = evaluate(model, val_loader, criterion)
    if va < best_val:
        best_val, best_state, best_epoch = va, copy.deepcopy(model.state_dict()), epoch

print("최적 에폭:", best_epoch, " 검증 손실:", round(best_val, 2))
print("마지막 검증 손실:", round(evaluate(model, val_loader, criterion), 2))
````

### 7.2 최적 지점으로 되돌리기

`load_state_dict`로 보관한 가중치를 모델에 다시 넣는다. 이것이 **조기 종료**다. 끝까지 돌리되 가장 좋았던 지점으로 되돌린다.

````python
model.load_state_dict(best_state)
print("되돌린 뒤 검증 손실:", round(evaluate(model, val_loader, criterion), 2))
````

검증 손실이 최적 에폭의 값(약 **5.05**)으로 돌아온다. 되돌리기 전의 마지막 검증 손실은 약 **9.5**였다.

## 8. 테스트 세트로 최종 평가

### 8.1 한 번만 본다

테스트 세트는 지금까지 한 번도 쓰지 않았다. 여기서 딱 한 번 평가한다.

````python
test_mse = evaluate(model, test_loader, criterion)
print("테스트 MSE :", round(test_mse, 2))
print("테스트 RMSE:", round(test_mse ** 0.5, 2))
````

테스트 RMSE는 약 **2.5**다. RMSE는 손실에 제곱근을 씌운 값이라 연비와 같은 단위다. 예측이 평균적으로 그만큼 빗나간다.

### 8.2 기준선과 비교

아무 모델 없이 훈련 세트의 평균 연비로만 예측하면 얼마나 틀리는지 구한다. 약 **7.1**이 나온다. 모델은 이 값보다 작아야 의미가 있다.

````python
baseline = ((y_te - y_tr.mean()) ** 2).mean().item()
print("기준선 RMSE:", round(baseline ** 0.5, 2))
````

## 9. 직접 해보기 ④ — 모델 크기 바꾸기

은닉층이 8개뿐인 작은 모델과 128·64개인 큰 모델을 7.1절과 같은 방식으로 학습시켜 최적 에폭과 최적 검증 손실을 비교한다. 아래 준비 셀은 그대로 실행한다.

````python
def train_with_best(model, epochs=200):
    optimizer = torch.optim.Adam(model.parameters(), lr=0.01)
    best_val, best_epoch = float('inf'), 0
    for epoch in range(epochs):
        train_one_epoch(model, train_loader, criterion, optimizer)
        va = evaluate(model, val_loader, criterion)
        if va < best_val:
            best_val, best_epoch = va, epoch
    return best_epoch, best_val
````

다음 셀의 주석을 코드로 바꾼다.

````python
# ✏️ 직접 채워 보세요
torch.manual_seed(42)
# ① small: Linear(9, 8) → ReLU → Linear(8, 1)
# ② big: Linear(9, 128) → ReLU → Linear(128, 64) → ReLU → Linear(64, 1)
# ③ 각 모델을 train_with_best로 학습시켜 (최적 에폭, 최적 검증 손실) 출력
````

| 모델 | 파라미터 수 | 최적 에폭 | 최적 검증 손실 |
|---|---|---|---|
| 작은 모델 (8) | 직접 기록 | 직접 기록 | 직접 기록 |
| 기본 모델 (32, 16) | 865 | 7.1절의 값 | 7.1절의 값 |
| 큰 모델 (128, 64) | 직접 기록 | 직접 기록 | 직접 기록 |

파라미터 수는 `sum(p.numel() for p in model.parameters())`로 센다. 큰 모델의 최적 에폭이 언제 오는지, 그 뒤 검증 손실이 어떻게 되는지 러닝커브로 확인한다.
